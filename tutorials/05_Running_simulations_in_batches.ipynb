# Running Simulations in Batches

MikFoil can run a batch of cases defined in a YAML file instead of configuring individual `Case` objects. In this tutorial you will get an idea how to quickly run batches of cases.

```python
from pathlib import Path
from mikfoil import cases_from_yaml

from mikfoil.main import execute_case


batch_yaml_path = Path("data/batch_cases.yaml")


with open(batch_yaml_path, "r") as f:
    print(f.read())
```

We can see that every case is defined with its parameters. We can modify those parameters either globally for all cases in batch. Or overwrite them specifically for a specific case.

The YAML file allows us to define global configurations (like `fineness: 1.5` and `su2_config: default`), but we can still override them on a per-case basis, as seen with the NACA 4412 using the `transition` configuration and `fineness: 2`.

```python

cases = cases_from_yaml(batch_yaml_path)


for case in cases:
    execute_case(case)
```

`cases_from_yaml` does nothing more, nothing less than creates a list of initialised `Case` objects. To run them all we can just use a for loop with an  `execute_case` function.

```python

for case in cases:
    print(f"{case.airfoil} - CL: {case.results.cl:.4f}, CD: {case.results.cd:.4f}")
```

So here is just another demonstration of how for loops allow for easy manipulation of many cases.

```python
from ipywidgets import Image, HBox


images = [Image(value=open(case.contours_dir / "contour_pressure.png", "rb").read()) for case in cases]
HBox(images)
```
